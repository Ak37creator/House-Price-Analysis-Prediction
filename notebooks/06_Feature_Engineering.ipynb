```python
# Load Libraries & dataset

import pandas as pd
import numpy as np

train_df = pd.read_csv("../data/processed/train_clean_for_eda.csv")
test_df = pd.read_csv("../data/processed/test_clean_for_eda.csv")

print("Training Shape:", train_df.shape)
print("Testing Shape:", test_df.shape)
```

```python
# Make House Age 

train_df["HouseAge"] = train_df["YrSold"] - train_df["YearBuilt"]

test_df["HouseAge"] = test_df["YrSold"] - test_df["YearBuilt"]

train_df[["YearBuilt", "YrSold", "HouseAge"]].head()
```

```python
# Make Remodeling Age 

train_df["RemodAge"] = train_df["YrSold"] - train_df["YearRemodAdd"]

test_df["RemodAge"] = test_df["YrSold"] - test_df["YearRemodAdd"]

train_df[["YearRemodAdd", "YrSold", "RemodAge"]].head()
```

```python
# Make Total Area of House (Different Different Sides)

train_df["TotalArea"] = (
    train_df["TotalBsmtSF"] +
    train_df["1stFlrSF"] +
    train_df["2ndFlrSF"]
)

test_df["TotalArea"] = (
    test_df["TotalBsmtSF"] +
    test_df["1stFlrSF"] +
    test_df["2ndFlrSF"]
)

train_df[[
    "TotalBsmtSF",
    "1stFlrSF",
    "2ndFlrSF",
    "TotalArea"
]].head()
```

```python
# Total Bathrooms Of House

train_df["TotalBathrooms"] = (
    train_df["FullBath"] +
    (0.5 * train_df["HalfBath"]) +
    train_df["BsmtFullBath"] +
    (0.5 * train_df["BsmtHalfBath"])
)

test_df["TotalBathrooms"] = (
    test_df["FullBath"] +
    (0.5 * test_df["HalfBath"]) +
    test_df["BsmtFullBath"] +
    (0.5 * test_df["BsmtHalfBath"])
)

train_df[[
    "FullBath",
    "HalfBath",
    "BsmtFullBath",
    "BsmtHalfBath",
    "TotalBathrooms"
]].head()
```

```python
# Total Porch Area Of House

porch_columns = [
    "OpenPorchSF",
    "EnclosedPorch",
    "3SsnPorch",
    "ScreenPorch",
    "WoodDeckSF"
]

train_df["TotalPorchArea"] = train_df[porch_columns].sum(axis=1)

test_df["TotalPorchArea"] = test_df[porch_columns].sum(axis=1)

train_df[porch_columns + ["TotalPorchArea"]].head()
```

```python
# Make 1 to 10 Total Quality Category

def quality_category(value):
    if value <= 4:
        return "Low"
    elif value <= 7:
        return "Medium"
    else:
        return "High"

train_df["QualityCategory"] = train_df["OverallQual"].apply(quality_category)

test_df["QualityCategory"] = test_df["OverallQual"].apply(quality_category)

train_df[["OverallQual", "QualityCategory"]].head(10)
```

```python
# Check News Features

new_features = [
    "HouseAge",
    "RemodAge",
    "TotalArea",
    "TotalBathrooms",
    "TotalPorchArea",
    "QualityCategory"
]

train_df[new_features].head()
```

```python
# Check missing values & negative values 

print("Training Missing Values:")
print(train_df[new_features].isnull().sum())

print("\nTesting Missing Values:")
print(test_df[new_features].isnull().sum())

print("\nNegative House Age:", (train_df["HouseAge"] < 0).sum())
print("Negative Remodel Age:", (train_df["RemodAge"] < 0).sum())
```

```python
# Save Processed Datasets 

train_df.to_csv(
    "../data/processed/train_feature_engineered.csv",
    index=False
)

test_df.to_csv(
    "../data/processed/test_feature_engineered.csv",
    index=False
)

print("Feature Engineering Completed Successfully!")
```
