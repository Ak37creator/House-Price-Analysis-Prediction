```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
train_clean = pd.read_csv("../data/processed/train_clean_for_eda.csv")
test_clean = pd.read_csv("../data/processed/test_clean_for_eda.csv")

# Display first 5 rows
train_clean.head()
```

```python
# get info of Database

print("Dataset Shape:", train_clean.shape)

print("\nFirst 5 Rows:")
display(train_clean.head())

print("\nDataset Information:")
train_clean.info()

print("\nStatistical Summary:")
train_clean.describe()
```

```python
# see SalePrice Distribution

plt.figure(figsize=(10, 5))

sns.histplot(
    train_clean["SalePrice"],
    bins=30,
    kde=True
)

plt.title("Distribution of House Sale Prices")
plt.xlabel("Sale Price")
plt.ylabel("Number of Houses")

plt.show()
```

```python
# Visualize Inside ofSalePrice & Outliers

plt.figure(figsize=(10, 4))

sns.boxplot(x=train_clean["SalePrice"])

plt.title("Boxplot of House Sale Prices")
plt.xlabel("Sale Price")

plt.show()
```

```python
# House Shape and Price Both Relation

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=train_clean,
    x="GrLivArea",
    y="SalePrice"
)

plt.title("Living Area vs Sale Price")
plt.xlabel("Ground Living Area (sq ft)")
plt.ylabel("Sale Price")

plt.show()
```

```python
# Overall Quality & SalePrice

quality_price = train_clean.groupby("OverallQual")["SalePrice"].mean()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=quality_price.index,
    y=quality_price.values
)

plt.title("Average Sale Price by Overall Quality")
plt.xlabel("Overall Quality")
plt.ylabel("Average Sale Price")

plt.show()
```

```python
# Neighborhood base House Pricing 

neighborhood_price = (
    train_clean.groupby("Neighborhood")["SalePrice"]
    .median()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=neighborhood_price.index,
    y=neighborhood_price.values
)

plt.title("Median House Price by Neighborhood")
plt.xlabel("Neighborhood")
plt.ylabel("Median Sale Price")

plt.xticks(rotation=90)

plt.show()
```

```python
# Features of correlation with Saleprice

numeric_data = train_clean.select_dtypes(include=np.number)

correlation = numeric_data.corr(numeric_only=True)

saleprice_corr = correlation["SalePrice"].sort_values(
    ascending=False
)

display(saleprice_corr)
```

```python
# 10 Features of More Positive Correlation With Saleprice

top_features = saleprice_corr.drop("SalePrice").head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_features.values,
    y=top_features.index
)

plt.title("Top 10 Features Correlated with SalePrice")
plt.xlabel("Correlation")
plt.ylabel("Features")

plt.show()
```
